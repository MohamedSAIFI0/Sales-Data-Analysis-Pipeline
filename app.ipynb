### Data Cleaning & Preprocessing [Customers]

```python
customers.dropna(inplace=True)
```

```python
customers['country'] = customers['country'].str.lower().str.strip()
```

```python
customers['name'] = customers['name'].str.strip().str.lower()
```

```python
customers['customer_id'] = customers['customer_id'].astype('int16')
```

```python
customers.columns
```

```python
customers.dtypes
```

```python
customers['customer_country'] = customers['customer_country'].astype('category')
```

```python
customers.rename(columns={
    "customer_id":"customer_id",
    "name":"customer_name",
    "country":"customer_country"
}, inplace=True)
```

```python
customers.columns
```

### Data Cleaning & Preprocessing [Products]

```python
products.dropna(inplace=True)
```

```python
products.columns
```

```python
products.rename(columns={
    "category":"product_category",
    "price_category":"product_price"
}, inplace=True)
```

```python
products.dtypes
```

```python
products['product_name'] = products['product_name'].str.strip().str.lower()
```

```python
products['product_category'] = products['product_category'].str.strip().str.lower()
```

```python
products.head()
```

```python
products['product_category'] = products['product_category'].astype("category")
```

```python
products.dtypes
```

```python
products["product_name"] = products["product_name"].str.replace(" ","_")
```
